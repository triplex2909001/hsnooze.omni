```python
# 4. Define Execution Paths on Google Drive
import os
from pathlib import Path

GDRIVE_BASE = Path('/content/drive/MyDrive')
CODEBASE_DIR = GDRIVE_BASE / '00.codebases'

PROJECT_NAME = "Matsuo Bashō - Matsuo Bashō - Whispering Pines and Quiet Trails: A Sleep Journey Through Edo Japan"

# Search candidates for the project folder on Google Drive
candidates = [
    GDRIVE_BASE / PROJECT_NAME,
    GDRIVE_BASE / 'historysnooze posts' / PROJECT_NAME,
]
TARGET_PROJECT_DIR = None
for c in candidates:
    if c.exists():
        TARGET_PROJECT_DIR = c
        break

if TARGET_PROJECT_DIR is None:
    matches = list(GDRIVE_BASE.glob(f"*{PROJECT_NAME}*"))
    if matches:
        TARGET_PROJECT_DIR = matches[0]

print(f'Target Project Directory: {TARGET_PROJECT_DIR}')
assert TARGET_PROJECT_DIR and TARGET_PROJECT_DIR.exists(), f'Project folder not found on Drive. Checked: {candidates}'
```
